## 1. My lane as an ML task

I chose the **Refresh / Content Opportunity Scoring** lane.

This is mainly a **scoring** problem. The goal is to give each content page a priority score based on its available signals, so pages that are more likely to need a refresh appear at the top of the list. Instead of predicting a category, the output is a score that helps decide which pages should be reviewed first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**My Answer**

The dataset does not have a column that directly says whether a page needs a refresh. Instead, I will use a combination of available signals such as trend direction, CTR, engagement, and days since the last update as a proxy.

The final output will be a priority score that helps identify pages that may benefit from a content review.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**My Answer**

A good result is a ranked list where the highest-scoring pages are reasonable candidates for a content refresh. The recommendations should make sense based on the available data and help a content team decide where to focus first.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is **one content page**.

Each row in the dataset represents one content page along with its search performance, engagement, content details, and freshness information. These features can be used together to estimate the page's refresh priority.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**My Answer**

A fixed rule might only look at one factor, such as content age or CTR. In reality, page performance depends on several signals that work together.

Machine learning can consider multiple features at the same time and assign a priority score based on the overall pattern. This makes the recommendations more flexible than using a single if-statement.
