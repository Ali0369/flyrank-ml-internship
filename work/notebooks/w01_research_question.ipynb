## 3. Quick look at the data (2-3 real numbers)


The starter dataset contains **30,000 content pages** with **44 features** describing search performance, content characteristics, and user engagement.

From the data, **174 pages** have not been updated for more than 180 days, while **16,262 pages** show a declining performance trend. The average click-through rate (CTR) across the dataset is **0.51**.

These numbers suggest that there are many pages with different performance patterns, making this dataset suitable for identifying which content should be prioritized for review or refresh.

```python
import pandas as pd
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')


print(f"Total Pages: {len(df):,}")
print(f"Total Features: {len(df.columns)}")

stale_pages = (df["days_since_last_update"] > 180).sum()
print(f"Pages not updated for more than 180 days: {stale_pages:,}")

declining_pages = (df["trend_direction"] == "down").sum()
print(f"Pages with declining trend: {declining_pages:,}")

print(f"Average CTR: {df['ctr'].mean():.2f}%")
```
